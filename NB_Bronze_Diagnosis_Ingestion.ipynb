```python

# Read diagnosis CSV from Azure Storage Landing
df = (
    spark.read
    .format("csv")
    .option("header", "true")
    .load("abfss://healthcare@ecommerceunifiedproject.dfs.core.windows.net/Landing/diagnosis_raw.csv")
)

# Write to the Unity Catalog Bronze schema
(
    df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("healthcare_dev.bronze.diagnosis_raw")
)
```
